```python
import pandas as pd

raw_path = r"E:\cyclocast_module1\data\raw\ibtracs.NI.list.v04r01.csv"
df = pd.read_csv(raw_path, low_memory=False)

print("Shape of dataset:", df.shape)
df.head()
```

```python
# See all column names
print(list(df.columns))
```

```python
# Check which basins exist in the data
print(df['BASIN'].unique())
```

```python
# Check how many rows belong to the North Indian Ocean basin
print(df[df['BASIN'] == 'NI'].shape)
```

```python
# Look for columns related to wind, pressure, and category
wind_pressure_cols = [col for col in df.columns if 'WIND' in col or 'PRES' in col or 'CAT' in col]
print(wind_pressure_cols)
```

```python
# Filter to NI basin only, and inspect New Delhi (IMD) columns
ni_df = df[df['BASIN'] == 'NI']

# Check unique values in NEWDELHI_GRADE (likely the category label)
print("NEWDELHI_GRADE unique values:")
print(ni_df['NEWDELHI_GRADE'].unique())

# Check how much data is missing in key columns
cols_to_check = ['ISO_TIME', 'LAT', 'LON', 'NEWDELHI_WIND', 'NEWDELHI_PRES', 'NEWDELHI_GRADE', 'WMO_WIND', 'WMO_PRES', 'SEASON', 'SID', 'NAME']
print("\nMissing value counts (out of", len(ni_df), "rows):")
print(ni_df[cols_to_check].isna().sum())

```

```python
# Check for blank/space values (not just NaN) in key columns
cols_to_check = ['NEWDELHI_WIND', 'NEWDELHI_PRES', 'NEWDELHI_GRADE', 'WMO_WIND', 'WMO_PRES']

for col in cols_to_check:
    blank_count = (ni_df[col].astype(str).str.strip() == '').sum()
    print(f"{col}: {blank_count} blank rows out of {len(ni_df)}")
```

```python
# Step 1: Drop the first row (row 0), which contains units, not real data
ni_df = ni_df.drop(index=0, errors='ignore')

# Step 2: Convert SEASON to numeric (this turns text like "1842" into the actual number 1842)
ni_df['SEASON'] = pd.to_numeric(ni_df['SEASON'], errors='coerce')

# Step 3: Check the year range again
print("Year range in NI data:", ni_df['SEASON'].min(), "to", ni_df['SEASON'].max())

# Step 4: Now filter to project's actual date range: 2013–2021
ni_recent = ni_df[(ni_df['SEASON'] >= 2013) & (ni_df['SEASON'] <= 2021)]
print("\nRows in 2013–2021 range:", ni_recent.shape)

# Step 5: Recheck blank counts, but only within 2013–2021
cols_to_check = ['NEWDELHI_WIND', 'NEWDELHI_PRES', 'NEWDELHI_GRADE', 'WMO_WIND', 'WMO_PRES']

for col in cols_to_check:
    blank_count = (ni_recent[col].astype(str).str.strip() == '').sum()
    print(f"{col}: {blank_count} blank rows out of {len(ni_recent)}")
```

```python
# Save the NI-filtered, 2013-2021 filtered raw data (not cleaned yet, just filtered)
output_path = r"E:\cyclocast_module1\data\raw\best_track_raw.csv"
ni_recent.to_csv(output_path, index=False)

print("Saved to:", output_path)
print("Final shape:", ni_recent.shape)
```

```python
import pandas as pd

raw_path = r"E:\cyclocast_module1\data\raw\best_track_raw.csv"
df = pd.read_csv(raw_path, low_memory=False)

print(df.shape)
df.head()
```

```python
# Keep only the essential columns
columns_needed = ['SID', 'NAME', 'SEASON', 'ISO_TIME', 'LAT', 'LON', 
                   'NEWDELHI_WIND', 'NEWDELHI_PRES', 'NEWDELHI_GRADE']

clean_df = df[columns_needed].copy()
print(clean_df.shape)
clean_df.head()
```

```python
clean_df = clean_df.rename(columns={
    'SID': 'storm_id',
    'NAME': 'storm_name',
    'SEASON': 'year',
    'ISO_TIME': 'timestamp',
    'LAT': 'lat',
    'LON': 'lon',
    'NEWDELHI_WIND': 'wind_speed',
    'NEWDELHI_PRES': 'pressure',
    'NEWDELHI_GRADE': 'category'
})

clean_df.head()
```

```python
clean_df['timestamp'] = pd.to_datetime(clean_df['timestamp'], errors='coerce')

print(clean_df['timestamp'].dtype)
clean_df[['storm_id', 'timestamp']].head()
```

```python
# Convert wind_speed and pressure to numeric (they may currently be text due to blank spaces)
clean_df['wind_speed'] = pd.to_numeric(clean_df['wind_speed'], errors='coerce')
clean_df['pressure'] = pd.to_numeric(clean_df['pressure'], errors='coerce')

# Clean up the category column: strip spaces, replace truly blank entries with NaN
clean_df['category'] = clean_df['category'].astype(str).str.strip()
clean_df['category'] = clean_df['category'].replace('', pd.NA)

clean_df.info()
```

```python
print("Missing timestamp:", clean_df['timestamp'].isna().sum())
print("Missing wind_speed:", clean_df['wind_speed'].isna().sum())
print("Missing pressure:", clean_df['pressure'].isna().sum())
print("Missing category:", clean_df['category'].isna().sum())
print("Total rows:", len(clean_df))
```

```python
# Final look before saving
print(clean_df.dtypes)
print("\nFinal shape:", clean_df.shape)
clean_df.head(10)
```

```python
clean_path = r"E:\cyclocast_module1\data\clean\best_track_clean.csv"
clean_df.to_csv(clean_path, index=False)

print("Saved to:", clean_path)
print("Final shape:", clean_df.shape)
```
